# TUPLE

```python
fruit = ("Apple","Mango","Grapes","Melon","Cherry")
print("First Elemnet:",fruit[1])
print("Last Elemnet:",fruit[-1])
new_list = list(fruit)
new_list[2] = "Orange"
print("Midfied List:",new_list)
```

# SET

```python
set1 ={1,2,10,20}
set2 ={1,40,10,45}
print("UNION",set1|set2)
print("Intersection",set1&set2)
print("Difference",set1-set2)
set1.add(15)
set2.add(20)
print("Updated set1:",set1)
print("Updated set2:",set2)
```
